```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
students = np.array([f"Student_{i}" for i in range(1, 51)])

print("=" * 55)
print("       STUDENT MARKS ANALYZER")
print("=" * 55)
```

```python
subjects = ["Maths", "Science", "English", "Computer", "Hindi"]
```

```python
np.random.seed(42)
marks = np.random.randint(35, 100, (50, 5))
```

```python
print("\nARRAY INFORMATION")
print("Shape:", marks.shape)
print("Dimensions:", marks.ndim)
print("Total elements:", marks.size)
print("Data type:", marks.dtype)

print("\nSUMMARY")
print("Total students:", marks.shape[0])
print("Total subjects:", marks.shape[1])
print("Maximum possible marks:", marks.shape[1] * 100)
```

```python
print("\nFIRST 10 STUDENTS' MARKS")
print(marks[:10])

print("\nLAST SUBJECT MARKS (HINDI)")
print(marks[:, -1])

print("\nFIRST THREE SUBJECTS")
print(marks[:, :3])

print("\nSUBJECTS 2 TO 4 (SCIENCE TO COMPUTER)")
print(marks[:, 1:4])

print("\nFIRST 5 STUDENTS, LAST 2 SUBJECTS")
print(marks[:5, -2:])
```

```python
subject_avg = np.mean(marks, axis=0)

plt.figure(figsize=(8, 5))
plt.bar(subjects, subject_avg)
plt.title("Subject-Wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.show()
```

```python
total_marks = np.sum(marks, axis=1)
percentage = total_marks / marks.shape[1]

print("\nTOTAL MARKS OF FIRST 10 STUDENTS")
print(total_marks[:10])
```

```python
print("\nSTUDENTS SCORING BELOW 40 IN ANY SUBJECT")
below_40_mask = np.any(marks < 40, axis=1)
print(students[below_40_mask])
print("Count:", np.sum(below_40_mask))

print("\nSTUDENTS SCORING ABOVE 90 IN ANY SUBJECT")
above_90_mask = np.any(marks > 90, axis=1)
print(students[above_90_mask])
print("Count:", np.sum(above_90_mask))
```

```python
grade_A = percentage >= 90
grade_B = (percentage >= 80) & (percentage < 90)
grade_C = (percentage >= 70) & (percentage < 80)
grade_D = (percentage >= 60) & (percentage < 70)
grade_E = (percentage >= 40) & (percentage < 60)
grade_F = percentage < 40

print("\nGRADE BAND COUNTS")
print("Grade A:", np.sum(grade_A))
print("Grade B:", np.sum(grade_B))
print("Grade C:", np.sum(grade_C))
print("Grade D:", np.sum(grade_D))
print("Grade E:", np.sum(grade_E))
print("Grade F:", np.sum(grade_F))
```

```python
print("\nSUBJECT-WISE TOPPERS")

for i in range(len(subjects)):
    highest = np.max(marks[:, i])
    topper_mask = marks[:, i] == highest

    print("\nSubject:", subjects[i])
    print("Highest marks:", highest)
    print("Topper(s):", students[topper_mask])
```

```python
pass_mask = np.all(marks >= 40, axis=1)
fail_mask = ~pass_mask

passed_students = students[pass_mask]
failed_students = students[fail_mask]

pass_percentage = np.mean(pass_mask) * 100
fail_percentage = np.mean(fail_mask) * 100

print("\nPASS AND FAIL REPORT")
print("Passed students:", len(passed_students))
print("Failed students:", len(failed_students))
print("Pass percentage:", round(pass_percentage, 2), "%")
print("Fail percentage:", round(fail_percentage, 2), "%")
```

```python
print("\nSUBJECT-WISE AVERAGES")
subject_avg = np.mean(marks, axis=0)

for subject, avg in zip(subjects, subject_avg):
    print(subject, ":", round(avg, 2))
```

```python
print("\n" + "=" * 55)
print("          FINAL PROJECT SUMMARY")
print("=" * 55)

print("Class average percentage:", round(np.mean(percentage), 2), "%")
print("Highest total marks:", np.max(total_marks), "/ 500")
print("Lowest total marks:", np.min(total_marks), "/ 500")
print("Pass percentage:", round(pass_percentage, 2), "%")
print("Fail percentage:", round(fail_percentage, 2), "%")
print("Most common grade:", max(
    ["A", "B", "C", "D", "E", "F"],
    key=lambda g: np.sum({
        "A": grade_A, "B": grade_B, "C": grade_C,
        "D": grade_D, "E": grade_E, "F": grade_F
    }[g])
))
```

```python
grade_names = ["A", "B", "C", "D", "E", "F"]

grade_counts = [
    np.sum(grade_A),
    np.sum(grade_B),
    np.sum(grade_C),
    np.sum(grade_D),
    np.sum(grade_E),
    np.sum(grade_F)
]

plt.figure(figsize=(8, 5))
plt.bar(grade_names, grade_counts)
plt.title("Grade Distribution of 50 Students")
plt.xlabel("Grades")
plt.ylabel("Number of Students")
plt.show()
```
